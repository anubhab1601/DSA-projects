## 🛠️ 1 · Setup

The scorer runs every solution in its own language, so we install the **six toolchains** the competition uses: 🐍 Python, 🟨 JavaScript (Node 22), ⚙️ C (gcc), 🐘 PHP, 💎 Ruby and 🌙 Lua. Python and gcc already come with Colab.

````python
#@title 🛠️ Install the six language toolchains (about 1 minute)
import os, sys, re, json, time, random, shutil, subprocess, zipfile, urllib.request, warnings, logging
os.environ['HF_HUB_DISABLE_IMPLICIT_TOKEN'] = '1'          # TigerCoder-1B is public: no Hugging Face token needed
os.environ['HF_HUB_VERBOSITY'] = 'error'; os.environ['TRANSFORMERS_VERBOSITY'] = 'error'
warnings.filterwarnings('ignore'); logging.getLogger('huggingface_hub').setLevel(logging.ERROR)

def sh(cmd):
    r = subprocess.run(cmd, shell=True, capture_output=True, text=True)
    if r.returncode:
        print(r.stdout[-1500:], r.stderr[-1500:]); raise RuntimeError(f'failed: {cmd}')
    return r.stdout.strip()

if not shutil.which('php') or not shutil.which('ruby') or not shutil.which('lua5.4'):
    print('📥 installing PHP, Ruby and Lua ...')
    sh('apt-get -qq update && DEBIAN_FRONTEND=noninteractive apt-get -qq install -y php-cli ruby lua5.4 > /dev/null')
sh('ln -sf "$(command -v lua5.4)" /usr/local/bin/lua')
if not sh('node --version 2>/dev/null || true').startswith('v22'):
    print('📥 installing Node 22 ...')
    sh('curl -fsSL https://nodejs.org/dist/v22.11.0/node-v22.11.0-linux-x64.tar.xz | tar -xJ -C /usr/local --strip-components=1')
# the scorer runs solutions with a fixed PATH (/usr/local/bin:/usr/bin:/bin); Colab keeps its Node in /tools/node
for tool in ['node', 'php', 'ruby', 'gcc']:
    where = shutil.which(tool)
    if where and not os.path.exists(f'/usr/local/bin/{tool}') and not where.startswith(('/usr/bin', '/bin')):
        os.symlink(where, f'/usr/local/bin/{tool}')

VERSIONS = {'🐍 Python': 'python3 --version', '🟨 Node': 'node --version', '⚙️ gcc': 'gcc --version | head -1',
            '🐘 PHP': 'php --version | head -1', '💎 Ruby': 'ruby --version', '🌙 Lua': 'lua -v'}
for name, cmd in VERSIONS.items():
    print(f'✅ {name:10s} {sh(cmd + " 2>&1")[:60]}')
````

### 🧭 Which programming language is it?

The data files do not name the programming language, but the signature line gives it away. A small rule reads it, and we check the rule on the trial set, whose reference knows the answer.

````python
#@title 🧭 Detect the programming language
PL_NAME = {'python': '🐍 Python', 'javascript': '🟨 JavaScript', 'c': '⚙️ C', 'php': '🐘 PHP', 'ruby': '💎 Ruby', 'lua': '🌙 Lua'}

def detect_pl(prompt):
    lines = [l for l in prompt.strip().splitlines() if l.strip()]
    if lines[0].startswith('def ') and lines[0].rstrip().endswith(':'):
        return 'python'                        # Python: signature first, docstring after
    last = lines[-1].strip()
    if last.startswith('def '):
        return 'ruby'
    if last.startswith('function '):
        if '$' in last or re.search(r'\)\s*:\s*\??\w+\s*\{$', last):
            return 'php'
        return 'javascript' if last.endswith('{') else 'lua'
    return 'c'

hits = sum(detect_pl(r['prompt']) == REF_TRIAL[r['task_id']]['pl'] for r in trial)
print(f'🎯 language detection on trial: {hits}/{len(trial)} correct')
counts = pd.Series([PL_NAME[detect_pl(r['prompt'])].split(' ', 1)[1] for r in dev]).value_counts()   # chart fonts have no emoji
ax = counts.plot.barh(figsize=(7, 2.6), color='#45c4a0'); ax.invert_yaxis(); ax.set_title('Dev tasks per programming language')
ax.spines[['top', 'right']].set_visible(False); plt.tight_layout(); plt.show()
````

## 📏 3 · The metric: Pass@1

$$\text{Pass@1} \;=\; \frac{\text{tasks whose solution passes \textbf{all} its test cases}}{\text{all tasks}}$$

* **One solution per task**, no partial credit: one failing test case fails the task. ❌
* Each task has **about 50 test cases**. The dev and test files show 2 and the trial file shows 5; the rest are hidden 🙈, and all are scored.
* Every solution runs in a sandbox: **1 s of CPU time**, 512 MB of memory, no network, no extra files, standard library only. Infinite loops are cut at 1 s. ⏱️
* The leaderboard also shows Pass@1 for each **language class** (Joshi et al., 2020) and each **programming language**; the ranking uses overall Pass@1. 🏆

The helper below runs the **official scorer** on a set of solutions. On the trial set it knows all the tests, so you get the exact score CodaBench would give.

````python
#@title 📏 The official scorer, as a helper
def score(solutions, tasks=None, workdir='run'):
    """Score {task_id: solution} on trial tasks with the official scorer. Returns (scores, outcome per task)."""
    tasks = tasks or list(REF_TRIAL)
    shutil.rmtree(workdir, ignore_errors=True)
    os.makedirs(f'{workdir}/ref'); os.makedirs(f'{workdir}/res')
    with open(f'{workdir}/ref/reference.jsonl', 'w') as f:
        f.writelines(json.dumps(REF_TRIAL[t]) + '\n' for t in tasks)
    json.dump({'phase': 'trial', 'detailed_results': True}, open(f'{workdir}/ref/phase.json', 'w'))
    with open(f'{workdir}/res/submission.jsonl', 'w') as f:
        f.writelines(json.dumps({'task_id': t, 'solution': solutions[t]}) + '\n' for t in tasks if t in solutions)
    r = subprocess.run([sys.executable, 'scorer/score.py', workdir, f'{workdir}/out'], capture_output=True, text=True)
    if r.returncode:
        raise RuntimeError(r.stdout + r.stderr)
    s = json.load(open(f'{workdir}/out/scores.json'))
    html_ = open(f'{workdir}/out/detailed_results.html').read()
    outcome = dict(re.findall(r'<tr><td>(T1-[a-z]+-\d+)</td><td class="\w+">([a-z ]+)</td></tr>', html_))
    return s, outcome

def card(title, value, sub='', dark=False):
    bg, fg, bd = ('#16202f', '#fff', '#16202f') if dark else ('#fff', '#16202f', '#d0d7de')
    return (f'<div style="padding:12px 18px;border-radius:14px;background:{bg};color:{fg};border:1px solid {bd};min-width:110px">'
            f'<div style="font-size:11px;letter-spacing:.04em;opacity:.7">{title}</div>'
            f'<div style="font-size:28px;font-weight:700">{value}</div><div style="font-size:11px;opacity:.6">{sub}</div></div>')

def show_scores(s, label):
    pls = ''.join(card(PL_NAME[p].split(' ', 1)[1].upper(), f'{s[p]:.1f}', PL_NAME[p].split(' ')[0]) for p in PL_NAME)
    cls = ''.join(card(f'CLASS {k}', f'{s[f"tier_{k}"]:.1f}') for k in range(4) if f'tier_{k}' in s)
    main = card('PASS@1 · ' + label, '%.1f%%' % s['score'], '', True)
    display(HTML(f'<div style="display:flex;flex-wrap:wrap;gap:10px;margin:8px 0">{main}{pls}{cls}</div>'))
````

````python
#@title ✅ Sanity check: the reference solutions score 100%
gold = {r['task_id']: r['canonical_solution'] for r in trial}
s, _ = score(gold); show_scores(s, 'reference solutions')
broken = [PL_NAME[p] for p in PL_NAME if s[p] < 100]
assert not broken, f'🚫 These toolchains do not work yet: {broken}. Run the setup cell again.'
print('✅ all six languages run: the scorer is ready')
s, _ = score({t: '' for t in gold}); print(f'🙃 empty submission → Pass@1 {s["score"]:.1f}%')
````

## 🐯 4 · Prompting TigerCoder-1B

[TigerCoder-1B](https://huggingface.co/md-nishat-008/TigerCoder-1B) is a compact code model (Gemma 3, 1B parameters) trained to write code from Bangla instructions. Here it meets 50 languages and six programming languages, many of them new to it. That is the challenge. 💪

The recipe is plain on purpose:

1. wrap the task prompt in a short **English instruction** that names the programming language;
2. ✍️ **start the answer for the model**: a code fence and the task prompt itself, so it continues *inside* the function. Left alone, TigerCoder often answers in Python, its strongest language, even when the task is in Lua or JavaScript;
3. let the model continue **greedily** (one attempt per task, as Pass@1 counts) until it closes the code block;
4. add a few common imports or headers, and drop demo code.

````python
#@title 🐯 Load TigerCoder-1B and define the prompt
from transformers import AutoTokenizer, AutoModelForCausalLM

MODEL_ID = 'md-nishat-008/TigerCoder-1B'
BATCH, MAX_NEW_TOKENS = 16, 512

tok = AutoTokenizer.from_pretrained(MODEL_ID)
tok.padding_side = 'left'                      # left padding for batched generation
from transformers.utils import logging as hf_logging
hf_logging.set_verbosity_error()
model = AutoModelForCausalLM.from_pretrained(MODEL_ID, dtype=DTYPE).cuda().eval()
print(f'🐯 {MODEL_ID}: {sum(p.numel() for p in model.parameters()) / 1e9:.2f}B parameters')

LANG = {'python': 'Python', 'javascript': 'JavaScript', 'c': 'C', 'php': 'PHP', 'ruby': 'Ruby', 'lua': 'Lua'}

def instruction(prompt, pl):
    return (f'Complete the following {LANG[pl]} function. The problem is described in the comment. '
            f'Reply with the complete function only, in one ```{pl}``` code block, '
            f'with no main function, no tests and no printing.\n\n```{pl}\n{prompt}```')

def start_of_answer(prompt, pl):
    """We write the start of the answer ourselves: a code fence and the task's own prompt. The model then
    continues inside the function, in the right language (left alone, TigerCoder drifts to Python)."""
    return f'```{pl}\n{prompt}'

@torch.inference_mode()
def generate(model, prompts, desc='generate'):
    """Greedy continuations for a list of (prompt, pl) pairs, in batches."""
    model.eval(); answers = []
    for i in tqdm(range(0, len(prompts), BATCH), desc=desc):
        texts = [tok.apply_chat_template([{'role': 'user', 'content': instruction(p, pl)}], tokenize=False,
                                         add_generation_prompt=True) + start_of_answer(p, pl)
                 for p, pl in prompts[i:i + BATCH]]
        enc = tok(texts, return_tensors='pt', padding=True, add_special_tokens=False).to(model.device)
        out = model.generate(**enc, max_new_tokens=MAX_NEW_TOKENS, do_sample=False, pad_token_id=tok.pad_token_id)
        answers += tok.batch_decode(out[:, enc['input_ids'].shape[1]:], skip_special_tokens=True)
    return answers

# common imports and headers the body may need: the answer starts inside the function, so it cannot add them
PREAMBLE = {'python': 'import math, re, heapq, bisect, string\nfrom collections import Counter, defaultdict, deque\n'
                      'from functools import lru_cache, reduce\nfrom itertools import *\n',
            'c': '#include <limits.h>\n#include <ctype.h>\n#include <stdint.h>\n',
            'ruby': "require 'set'\n"}

def to_solution(answer, row, pl):
    """Prompt + the model's continuation up to the closing fence, cleaned up."""
    cont = answer.split('```')[0]
    if pl == 'python':                                       # the body must be indented: drop stray text before it
        lines = cont.split('\n')
        while lines and lines[0].strip() and not lines[0].startswith((' ', '\t')):
            lines.pop(0)
        cont = '\n'.join(lines)
    code = row['prompt'] + cont
    if pl == 'python':
        code = code.split('if __name__')[0]                 # drop demo code that would run
    if pl == 'c':
        code = re.split(r'\n\s*int\s+main\s*\(', code)[0]    # the scorer brings its own main
    if pl == 'php':
        code = re.sub(r'\?>\s*$', '', code.rstrip())
    return PREAMBLE.get(pl, '') + code.strip() + '\n'
````

## 🧪 5 · Step by step on trial

**Step 5a** asks the model to solve all 100 trial tasks. **Step 5b** turns the answers into solutions. **Step 5c** scores them with the official scorer, just as CodaBench would.

````python
#@title 🧪 5a · Ask the model for every trial task
t0 = time.time()
trial_answers = generate(model, [(r['prompt'], detect_pl(r['prompt'])) for r in trial], 'trial')
print(f'⏱️ {len(trial)} answers in {time.time() - t0:.0f} s')
print('─' * 70, '\n🗣️ What the model wrote after the prompt we gave it:\n'); print(trial_answers[0][:1200])
````

````python
#@title 🧹 5b · Turn the answers into solutions
trial_solutions = {r['task_id']: to_solution(a, r, detect_pl(r['prompt'])) for r, a in zip(trial, trial_answers)}
print('🧹 Prompt + continuation, cleaned into a solution:\n'); print(trial_solutions[trial[0]['task_id']])
````

````python
#@title 📊 5c · Score the trial solutions
s_trial, outcome = score(trial_solutions)
show_scores(s_trial, 'TigerCoder-1B, trial')
oc = pd.Series(outcome).value_counts()
ax = oc.plot.barh(figsize=(7, 2.4), color=['#2e7d32' if k == 'passed' else '#c62828' for k in oc.index])
ax.invert_yaxis(); ax.set_title('Outcome of each trial task'); ax.spines[['top', 'right']].set_visible(False)
plt.tight_layout(); plt.show()
````

## 🔬 6 · What passed, what failed?

Reading failures is the fastest way to improve. Each outcome has a reason: **wrong answer** (the logic or the output format), **runtime error** (a crash), **compile error** (C), **timeout** (over 1 s of CPU).

````python
#@title 🔬 Show one passed and two failed tasks
def show_task(tid, sol):
    r = next(x for x in trial if x['task_id'] == tid)
    color = '#2e7d32' if outcome[tid] == 'passed' else '#c62828'
    display(HTML(f"""<div style="border-left:5px solid {color};padding:6px 14px;margin:10px 0">
    <b>{tid}</b> · {PL_NAME[detect_pl(r['prompt'])]} · <span style="color:{color}"><b>{outcome[tid]}</b></span>
    <details><summary>📝 prompt</summary><pre style="white-space:pre-wrap;font-size:12px">{r['prompt']}</pre></details>
    <details open><summary>🤖 solution</summary><pre style="font-size:12px">{sol}</pre></details>
    <details><summary>✅ reference</summary><pre style="font-size:12px">{r['canonical_solution']}</pre></details></div>"""))

passed = [t for t, o in outcome.items() if o == 'passed']
failed = [t for t, o in outcome.items() if o != 'passed']
for t in passed[:1] + failed[:2]:
    show_task(t, trial_solutions[t])
````

🔭 **Small model, small sample.** These scores come from TigerCoder-1B, a model with 1 billion parameters, given a plain prompt and measured on 100 trial tasks. Larger models will very likely score higher. Treat this run as the starting point to beat.

### 💡 Where to go from here

<details>
<summary><b>💡 Ideas worth a try (open when you are ready)</b></summary>

* 🧪 **Use the two visible tests.** Every dev and test task shows 2 of its tests. Sample several answers (`do_sample=True`), run them on the visible tests, and keep one that passes. The harness in `scorer/lc_harness.py` can run a task once you give it a signature; the types can be read from the test values.
* 🔁 **Repair loop.** When a solution fails a visible test, show the model the error or the wrong output and ask again.
* 📚 **Few-shot.** Put one or two solved trial tasks in the same programming language into the prompt.
* 🌍 **Language.** The descriptions come in 50 languages and TigerCoder knows Bangla and English best. Translating the description first can help, but keep every quoted text and every argument name exactly as written: the code must match them.
* 🧩 **Conventions.** Lua tables start at 1; C returns `malloc`'d arrays and sets `*out_len`; PHP arrays are re-indexed. Say so in the instruction.
* ✂️ **Length.** `MAX_NEW_TOKENS` cuts long answers; a cut-off answer has no closing code fence.
* 🔁 **No repetition penalty.** `repetition_penalty > 1` punishes every token already in the context, and the prompt is full of the argument names, keywords and indentation the code must reuse. We leave it at 1.0.
* 🦾 **Bigger models.** Any model or tool is allowed, as long as your system paper describes it.
</details>

## 🧯 Common errors

| What you see | Why | Fix |
|---|---|---|
| 🚫 *No GPU found* | The runtime has no GPU | Runtime → Change runtime type → T4 GPU |
| 💥 *CUDA out of memory* | Batch too large | Lower `BATCH` and run the generation cell again |
| ❌ Status **Failed** on CodaBench | The file broke a format rule | Run the format check cell and fix what it reports |
| 🙈 Many *missing* tasks | Solutions not written for every `task_id` | Write one line for every dev task |
| ⏱️ Many *timeout* outcomes | Slow or looping code | Each task gets 1 s of CPU time; check loops and recursion |
| ⚙️ C tasks fail to compile | A `main` function, or a different signature | Keep the signature of the prompt; no `main` |
| 🌙 Lua or ⚙️ C answers look wrong | Language conventions | Lua tables start at 1; C returns a `malloc`'d array and sets `*out_len`, as the prompt's signature shows |

````python
# ==============================================================================
# CELL 2: STOCHASTIC SAMPLING (EXACT BASELINE PROMPT) & CANDIDATE POOL AUDIT
# ==============================================================================
SEED = 42
random.seed(SEED)
np.random.seed(SEED)
torch.manual_seed(SEED)
if torch.cuda.is_available():
    torch.cuda.manual_seed_all(SEED)

# Correction 2: Set cuDNN deterministic flags for maximum reproducibility across runs.
# Note: This improves reproducibility on a given system but does not guarantee
# bitwise-identical GPU sampling across differing CUDA architectures, drivers, or library builds.
torch.backends.cudnn.deterministic = True
torch.backends.cudnn.benchmark = False

NUM_SAMPLES = 5
TEMP = 0.6
TOP_P = 0.9
MAX_NEW_TOKENS = 512

all_candidates = {}

for task in tqdm(trial_tasks, desc="Sampling candidates"):
    tid = task["task_id"]
    pl = trial_metadata[tid]["pl"]

    # Candidate 0: Exact immutable greedy baseline solution from Experiment 0
    candidates = [exp0_solutions[tid]]

    # EXACT ORIGINAL BASELINE PROMPT CONSTRUCTION (String-concatenation pattern)
    prompt_str = tok.apply_chat_template(
        [{"role": "user", "content": instruction(task["prompt"], pl)}],
        tokenize=False,
        add_generation_prompt=True
    ) + start_of_answer(task["prompt"], pl)

    input_ids = tok(prompt_str, return_tensors="pt")["input_ids"].to(model.device)
    prompt_len = input_ids.shape[1]

    # Generate 5 sampled candidates sequentially for T4 VRAM safety
    for _ in range(NUM_SAMPLES):
        with torch.no_grad():
            output_tokens = model.generate(
                input_ids,
                do_sample=True,
                temperature=TEMP,
                top_p=TOP_P,
                max_new_tokens=MAX_NEW_TOKENS,
                pad_token_id=tok.eos_token_id
            )
        raw_ans = tok.decode(output_tokens[0][prompt_len:], skip_special_tokens=True)
        # Apply exact official notebook solution cleaner
        cleaned = to_solution(raw_ans, task, pl)
        candidates.append(cleaned)

    if torch.cuda.is_available():
        torch.cuda.empty_cache()

    all_candidates[tid] = candidates

# Strict Post-Generation Validation
assert len(all_candidates) == 100, f"Expected 100 tasks in candidate pool, found {len(all_candidates)}"
for tid in trial_task_ids_ordered:
    cands = all_candidates[tid]
    assert len(cands) == 6, f"Task {tid} does not have exactly 6 candidates (found {len(cands)})"
    assert cands[0] == exp0_solutions[tid], f"Task {tid} Candidate 0 does not match Experiment 0 baseline!"
    for idx, c in enumerate(cands):
        assert isinstance(c, str) and c.strip(), f"Task {tid} Candidate {idx} is empty or invalid!"

print("✓ Cell 2 Complete:")
print("  - Configured deterministic seeds and cuDNN flags.")
print("  - Generated 5 sampled candidates per task with exact baseline prompt.")
print("  - Verified 6 candidates per task across all 100 tasks.")
print("  - Verified Candidate 0 is identical to Experiment 0 baseline.")
````

````python
# ==============================================================================
# CELL 3: VISIBLE-TEST FILTERING, CONSERVATIVE SELECTION & ARTIFACT PERSISTENCE
# ==============================================================================
selected_solutions = {}
eval_audit = {}
greedy_preserved = 0
sampled_promoted = 0

for task in tqdm(trial_tasks, desc="Filtering against visible tests"):
    tid = task["task_id"]
    min_row = trial_metadata[tid]
    visible_tests = task["tests"]
    candidates = all_candidates[tid]

    scores = []
    for idx, sol in enumerate(candidates):
        try:
            # 3.0 s wall clock timeout in harness
            oks, info = H.run(min_row, visible_tests, sol, timeout=3.0)
            passed = sum(1 for ok in oks if ok is True)
            err = info.get("error", None)
        except Exception as e:
            passed = 0
            err = str(e)

        scores.append({
            "idx": idx,
            "passed": passed,
            "error": err,
            "solution": sol
        })

    greedy_passed = scores[0]["passed"]
    max_passed = max(s["passed"] for s in scores)

    # Conservative Selection Rule:
    # If greedy achieved the maximum score (even if tied), strictly prefer greedy
    if greedy_passed == max_passed:
        chosen = scores[0]
        greedy_preserved += 1
    else:
        # Strictly promote the first sampled candidate achieving max_score
        chosen = next(s for s in scores[1:] if s["passed"] == max_passed)
        sampled_promoted += 1

    selected_solutions[tid] = chosen["solution"]
    eval_audit[tid] = {
        "greedy_passed": greedy_passed,
        "max_passed": max_passed,
        "chosen_idx": chosen["idx"],
        "scores": [{"idx": s["idx"], "passed": s["passed"], "error": s["error"]} for s in scores]
    }

# Ensure directory structures exist
os.makedirs("run_exp1/res", exist_ok=True)
os.makedirs("run_exp1/audit", exist_ok=True)
os.makedirs("evaluation/exp1_logs", exist_ok=True)

# 1. Save selected solutions for official scoring
with open("run_exp1/res/submission.jsonl", "w", encoding="utf-8") as f:
    for tid in trial_task_ids_ordered:
        f.write(json.dumps({"task_id": tid, "solution": selected_solutions[tid]}, ensure_ascii=False) + "\n")

# 2. Save full candidate pool (all 6 candidates per task)
with open("evaluation/exp1_logs/all_candidates_pool.json", "w", encoding="utf-8") as f:
    json.dump(all_candidates, f, indent=2, ensure_ascii=False)

# 3. Save visible-test evaluation audit
with open("evaluation/exp1_logs/visible_test_eval_results.json", "w", encoding="utf-8") as f:
    json.dump(eval_audit, f, indent=2, ensure_ascii=False)
shutil.copy("evaluation/exp1_logs/visible_test_eval_results.json", "run_exp1/audit/candidate_eval_records.json")

# 4. Save experiment configuration
exp1_config = {
    "experiment_id": "EXP-01",
    "description": "5 sampled candidates + visible test filtering with conservative greedy preference",
    "model": "md-nishat-008/TigerCoder-1B",
    "seed": SEED,
    "num_sampled": NUM_SAMPLES,
    "temperature": TEMP,
    "top_p": TOP_P,
    "max_new_tokens": MAX_NEW_TOKENS,
    "visible_test_timeout_sec": 3.0,
    "selection_rule": "Highest visible pass count; prefer Candidate 0 (greedy) on tie; never downgrade"
}
with open("evaluation/exp1_logs/exp1_config.json", "w", encoding="utf-8") as f:
    json.dump(exp1_config, f, indent=2)

print("\n--- Visible-Test Selection Summary ---")
print(f"Tasks where Greedy was preserved: {greedy_preserved}/100")
print(f"Tasks where Sampled outperformed Greedy: {sampled_promoted}/100")
print("✓ Cell 3 Complete: All artifacts saved in run_exp1/ and evaluation/exp1_logs/ (run/ is untouched).")
````

````python
# ==============================================================================
# EXPERIMENT 2 - CELL 2: SEQUENTIAL REPAIR GENERATION & CANDIDATE EVALUATION
# ==============================================================================
SEED = 42
random.seed(SEED)
np.random.seed(SEED)
torch.manual_seed(SEED)
if torch.cuda.is_available():
    torch.cuda.manual_seed_all(SEED)

torch.backends.cudnn.deterministic = True
torch.backends.cudnn.benchmark = False

TEMP = 0.2
TOP_P = 0.9
MAX_NEW_TOKENS = 512
MAX_ATTEMPTS = 2

def sanitize_json(obj):
    """Recursively converts objects to JSON-serializable primitives."""
    if obj is None or isinstance(obj, (bool, int, float, str)):
        return obj
    if isinstance(obj, dict):
        return {str(k): sanitize_json(v) for k, v in obj.items()}
    if isinstance(obj, (list, tuple)):
        return [sanitize_json(x) for x in obj]
    return str(obj)

def extract_compact_diagnostics(info):
    """Extracts a compact, JSON-safe diagnostic record from H.run() info."""
    return {
        "error": sanitize_json(info.get("error", None)),
        "first_failure": sanitize_json(info.get("first_failure", None)),
        "detail": sanitize_json(info.get("detail", None)),
        "stderr": sanitize_json(info.get("stderr", None)),
        "run_ms": sanitize_json(info.get("run_ms", None)),
        "compile_ms": sanitize_json(info.get("compile_ms", None))
    }

def build_diagnostic_feedback(info, pl):
    """Formats structured diagnostic feedback from H.run() output safely."""
    err = info.get("error", "")
    bad = info.get("first_failure", None)

    # 1. Compilation errors (primarily C)
    if err == "compile error":
        raw_detail = info.get("detail") if info.get("detail") is not None else info.get("stderr")
        detail_str = str(raw_detail) if raw_detail is not None else "Syntax or compilation error"
        detail = detail_str.strip()[-400:]
        return (f"The code failed to compile with the following compiler error:\n{detail}\n"
                f"Please fix the compilation errors. Provide only the complete corrected function in {pl}. "
                f"Do not write explanations, markdown commentary, or main functions.")

    # 2. Runtime errors / exceptions
    if err == "runtime error":
        raw_stderr = info.get("stderr")
        stderr_str = str(raw_stderr) if raw_stderr is not None else "Execution terminated with non-zero exit code."
        stderr = stderr_str.strip()[-400:]
        inp = str(bad.get("input")) if bad else "visible tests"
        return (f"The code encountered a runtime error during execution on test input: {inp}\n"
                f"Error trace:\n{stderr}\n"
                f"Please fix the error. Provide only the complete corrected function in {pl}. "
                f"Do not write explanations, markdown commentary, or main functions.")

    # 3. Timeout
    if err == "timeout":
        inp = str(bad.get("input")) if bad else "visible tests"
        return (f"The code exceeded the execution time limit (potential infinite loop or inefficient logic) on input: {inp}.\n"
                f"Please optimize the algorithm. Provide only the complete corrected function in {pl}. "
                f"Do not write explanations, markdown commentary, or main functions.")

    # 4. Wrong Answer / Assertion mismatch
    if bad is not None:
        inp = str(bad.get("input"))
        exp = str(bad.get("expected"))
        got = str(bad.get("got", "no output"))
        return (f"The code produced an incorrect result for test input: {inp}\n"
                f"Expected return: {exp}\n"
                f"Actual return: {got}\n"
                f"Please correct the logic. Provide only the complete corrected function in {pl}. "
                f"Do not write explanations, markdown commentary, or main functions.")

    # 5. Missing / extra output fallback
    return (f"The code did not produce the expected outputs on the visible test cases.\n"
            f"Please review the problem specifications and provide only the complete corrected function in {pl}. "
            f"Do not write explanations, markdown commentary, or main functions.")

def generate_repair_candidate(task, pl, previous_code, feedback_str):
    """Constructs multi-turn prompt without truncation and checks context limit."""
    messages = [
        {"role": "user", "content": instruction(task["prompt"], pl)},
        {"role": "assistant", "content": previous_code},
        {"role": "user", "content": feedback_str}
    ]

    # Apply chat template and append canonical start_of_answer prefix
    prompt_str = tok.apply_chat_template(
        messages,
        tokenize=False,
        add_generation_prompt=True
    ) + start_of_answer(task["prompt"], pl)

    # Tokenize WITHOUT truncation
    inputs = tok(prompt_str, return_tensors="pt")
    input_ids = inputs["input_ids"]
    prompt_len = input_ids.shape[1]

    # Check context budget
    if prompt_len + MAX_NEW_TOKENS > MODEL_CONTEXT_LIMIT:
        return None, f"Context overflow: prompt ({prompt_len} tokens) + max_new ({MAX_NEW_TOKENS}) > limit ({MODEL_CONTEXT_LIMIT})"

    try:
        input_ids = input_ids.to(model.device)
        with torch.no_grad():
            output_tokens = model.generate(
                input_ids,
                do_sample=True,
                temperature=TEMP,
                top_p=TOP_P,
                max_new_tokens=MAX_NEW_TOKENS,
                pad_token_id=tok.eos_token_id
            )
        raw_ans = tok.decode(output_tokens[0][prompt_len:], skip_special_tokens=True)
        sol = to_solution(raw_ans, task, pl)
        return sol, None
    except Exception as e:
        return None, f"Generation exception: {str(e)}"
    finally:
        if torch.cuda.is_available():
            torch.cuda.empty_cache()

# Tracking structures
all_candidates_pool = {}
repair_eval_audit = {}

for task in tqdm(trial_tasks, desc="Iterative Code Repair"):
    tid = task["task_id"]
    min_row = trial_metadata[tid]
    pl = min_row["pl"]
    visible_tests = task["tests"]

    candidates = []
    eval_history = []

    # --- CANDIDATE 0: Baseline Greedy Solution ---
    cand0 = exp0_solutions[tid]
    candidates.append(cand0)

    try:
        oks0, info0 = H.run(min_row, visible_tests, cand0, timeout=3.0)
        score0 = sum(1 for ok in oks0 if ok is True)
    except Exception as e:
        score0 = 0
        info0 = {"error": "runtime error", "stderr": str(e)}

    diag0 = extract_compact_diagnostics(info0)
    eval_history.append({
        "candidate_idx": 0,
        "passed": score0,
        "total": len(visible_tests),
        "diagnostics": diag0,
        "feedback_given": None,
        "generation_status": "baseline"
    })

    # If Candidate 0 passes 5/5 visible tests, skip repair entirely
    if score0 == len(visible_tests):
        all_candidates_pool[tid] = candidates
        repair_eval_audit[tid] = {
            "attempts_made": 0,
            "eval_history": eval_history
        }
        continue

    # --- REPAIR ATTEMPT 1: Generate Candidate 1 from Candidate 0 ---
    feedback1 = build_diagnostic_feedback(info0, pl)
    cand1, err1 = generate_repair_candidate(task, pl, cand0, feedback1)

    if cand1 is not None:
        candidates.append(cand1)
        try:
            oks1, info1 = H.run(min_row, visible_tests, cand1, timeout=3.0)
            score1 = sum(1 for ok in oks1 if ok is True)
        except Exception as e:
            score1 = 0
            info1 = {"error": "runtime error", "stderr": str(e)}

        diag1 = extract_compact_diagnostics(info1)
        eval_history.append({
            "candidate_idx": 1,
            "passed": score1,
            "total": len(visible_tests),
            "diagnostics": diag1,
            "feedback_given": feedback1,
            "generation_status": "success"
        })

        # If Candidate 1 passes 5/5, skip Attempt 2
        if score1 == len(visible_tests):
            all_candidates_pool[tid] = candidates
            repair_eval_audit[tid] = {
                "attempts_made": 1,
                "eval_history": eval_history
            }
            continue

        prev_cand_for_att2 = cand1
        prev_info_for_att2 = info1
    else:
        # Candidate 1 failed to generate (e.g. context overflow)
        eval_history.append({
            "candidate_idx": 1,
            "passed": -1,
            "total": len(visible_tests),
            "diagnostics": None,
            "feedback_given": feedback1,
            "generation_status": f"failed: {err1}"
        })
        # Cannot progress to Attempt 2 from Candidate 1; finish task
        all_candidates_pool[tid] = candidates
        repair_eval_audit[tid] = {
            "attempts_made": 1,
            "eval_history": eval_history
        }
        continue

    # --- REPAIR ATTEMPT 2: Generate Candidate 2 from Candidate 1 ---
    feedback2 = build_diagnostic_feedback(prev_info_for_att2, pl)
    cand2, err2 = generate_repair_candidate(task, pl, prev_cand_for_att2, feedback2)

    if cand2 is not None:
        candidates.append(cand2)
        try:
            oks2, info2 = H.run(min_row, visible_tests, cand2, timeout=3.0)
            score2 = sum(1 for ok in oks2 if ok is True)
        except Exception as e:
            score2 = 0
            info2 = {"error": "runtime error", "stderr": str(e)}

        diag2 = extract_compact_diagnostics(info2)
        eval_history.append({
            "candidate_idx": 2,
            "passed": score2,
            "total": len(visible_tests),
            "diagnostics": diag2,
            "feedback_given": feedback2,
            "generation_status": "success"
        })
    else:
        eval_history.append({
            "candidate_idx": 2,
            "passed": -1,
            "total": len(visible_tests),
            "diagnostics": None,
            "feedback_given": feedback2,
            "generation_status": f"failed: {err2}"
        })

    all_candidates_pool[tid] = candidates
    repair_eval_audit[tid] = {
        "attempts_made": 2,
        "eval_history": eval_history
    }

print("✓ Cell 2 Complete: Iterative repair passes executed with full context checking and error logging.")
````

````python
# ==============================================================================
# EXPERIMENT 2 - CELL 3: CONSERVATIVE SELECTION & ARTIFACT PERSISTENCE
# ==============================================================================
selected_solutions = {}
selection_summary = {
    "candidate_0_selected": 0,
    "candidate_1_selected": 0,
    "candidate_2_selected": 0,
    "strictly_improved_tasks": 0,
    "repaired_tasks_total": 0,
    "generation_skips_or_errors": 0
}

for tid in trial_task_ids_ordered:
    history = repair_eval_audit[tid]["eval_history"]
    cands = all_candidates_pool[tid]

    # Filter only successfully evaluated candidates (passed >= 0)
    valid_evals = [h for h in history if h["passed"] >= 0]

    score0 = valid_evals[0]["passed"]
    max_score = max(h["passed"] for h in valid_evals)

    # Conservative Selection Rule:
    # Tie-breaking priority: Candidate 0 > Candidate 1 > Candidate 2
    # Invariant: A repaired candidate is selected only if its score strictly exceeds score0
    if score0 == max_score:
        chosen_idx = 0
        selection_summary["candidate_0_selected"] += 1
    else:
        chosen_candidate = next(h for h in valid_evals[1:] if h["passed"] == max_score)
        chosen_idx = chosen_candidate["candidate_idx"]
        selection_summary[f"candidate_{chosen_idx}_selected"] += 1
        selection_summary["strictly_improved_tasks"] += 1

    selected_solutions[tid] = cands[chosen_idx]
    repair_eval_audit[tid]["selected_idx"] = chosen_idx
    repair_eval_audit[tid]["baseline_score"] = score0
    repair_eval_audit[tid]["final_visible_score"] = max_score

    if repair_eval_audit[tid]["attempts_made"] > 0:
        selection_summary["repaired_tasks_total"] += 1
    if any("failed" in h.get("generation_status", "") for h in history):
        selection_summary["generation_skips_or_errors"] += 1

# Setup dedicated output directories (strictly isolated from run/ and run_exp1/)
os.makedirs("run_exp2/res", exist_ok=True)
os.makedirs("run_exp2/audit", exist_ok=True)
os.makedirs("evaluation/exp2_logs", exist_ok=True)

# 1. Save selected solutions for official scoring
with open("run_exp2/res/submission.jsonl", "w", encoding="utf-8") as f:
    for tid in trial_task_ids_ordered:
        f.write(json.dumps({"task_id": tid, "solution": selected_solutions[tid]}, ensure_ascii=False) + "\n")

# 2. Save full candidate pool (all generated solutions per task)
with open("evaluation/exp2_logs/all_candidates_pool.json", "w", encoding="utf-8") as f:
    json.dump(all_candidates_pool, f, indent=2, ensure_ascii=False)

# 3. Save visible-test evaluation audit
with open("evaluation/exp2_logs/repair_eval_audit.json", "w", encoding="utf-8") as f:
    json.dump(repair_eval_audit, f, indent=2, ensure_ascii=False)
shutil.copy("evaluation/exp2_logs/repair_eval_audit.json", "run_exp2/audit/repair_eval_audit.json")

# 4. Save experiment configuration
exp2_config = {
    "experiment_id": "EXP-02",
    "description": "Iterative code repair (up to 2 sequential passes) with H.run visible-test diagnostics",
    "model": "md-nishat-008/TigerCoder-1B",
    "seed": SEED,
    "temperature": TEMP,
    "top_p": TOP_P,
    "max_new_tokens": MAX_NEW_TOKENS,
    "max_repair_attempts": MAX_ATTEMPTS,
    "model_context_limit": MODEL_CONTEXT_LIMIT,
    "visible_test_timeout_sec": 3.0,
    "selection_rule": "Highest visible pass count; tie priority Candidate 0 > Candidate 1 > Candidate 2"
}
with open("evaluation/exp2_logs/exp2_config.json", "w", encoding="utf-8") as f:
    json.dump(exp2_config, f, indent=2)

print("\n--- Visible-Test Selection Summary (Experiment 2) ---")
print(f"Total tasks evaluated for repair:      {selection_summary['repaired_tasks_total']}/100")
print(f"Tasks retaining Candidate 0 (Greedy):  {selection_summary['candidate_0_selected']}/100")
print(f"Tasks promoted to Candidate 1:         {selection_summary['candidate_1_selected']}/100")
print(f"Tasks promoted to Candidate 2:         {selection_summary['candidate_2_selected']}/100")
print(f"Total tasks strictly improved:         {selection_summary['strictly_improved_tasks']}/100")
print(f"Tasks with generation skips/errors:    {selection_summary['generation_skips_or_errors']}/100")
print("✓ Cell 3 Complete: All outputs safely saved under run_exp2/ and evaluation/exp2_logs/.")
````
